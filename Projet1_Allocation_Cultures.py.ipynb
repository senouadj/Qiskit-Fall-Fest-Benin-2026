```python
# ============================================================
# PROJET 1 : Allocation optimale de cultures au Bénin
# ============================================================

from qiskit_optimization import QuadraticProgram
from qiskit_optimization.algorithms import MinimumEigenOptimizer
from qiskit_algorithms import NumPyMinimumEigensolver, QAOA
from qiskit_algorithms.optimizers import COBYLA
from qiskit.primitives import StatevectorSampler
import numpy as np

print("Bibliothèques chargées avec succès !")

# --- 1. Définition du problème béninois ---
qp = QuadraticProgram(name="Allocation_Cultures_Benin")

qp.binary_var(name="Mais")
qp.binary_var(name="Manioc")
qp.binary_var(name="Coton")
qp.binary_var(name="Arachide")

# Maximiser le rendement (tonnes)
qp.maximize(
    linear={
        "Mais": 3.5,
        "Manioc": 12.0,
        "Coton": 1.8,
        "Arachide": 1.2
    }
)

# Contrainte : on ne plante que 2 cultures maximum
qp.linear_constraint(
    linear={"Mais": 1, "Manioc": 1, "Coton": 1, "Arachide": 1},
    sense="<=",
    rhs=2,
    name="surface_limitee"
)

print("\nProblème formulé :")
print(qp.prettyprint())

# --- 2. Solution exacte (classique) ---
exact_solver = MinimumEigenOptimizer(NumPyMinimumEigensolver())
result = exact_solver.solve(qp)

print("\n=== SOLUTION OPTIMALE ===")
print(result.prettyprint())

# --- 3. Interprétation claire ---
print("\n" + "="*55)
print("RECOMMANDATION POUR L'AGRICULTEUR BÉNINOIS")
print("="*55)

noms = ["Mais", "Manioc", "Coton", "Arachide"]
for nom, val in zip(noms, result.x):
    statut = "OUI, planter" if val > 0.5 else "NON"
    print(f"  {nom:12} : {statut}")

print(f"\nRendement total estimé : {result.fval:.1f} tonnes")
print("="*55)
print("Projet 1 terminé avec succès !")
```
